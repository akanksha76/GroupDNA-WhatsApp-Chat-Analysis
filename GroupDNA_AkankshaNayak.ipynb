# Project Overview

This project analyzes a WhatsApp chat exported as a text file.

The notebook extracts messages, detects user activity patterns,
finds the most active participants, computes word frequencies,
builds a NumPy activity heatmap, identifies personality archetypes,
and prints a formatted analytics report.

```python
import numpy as np

from collections import defaultdict

from collections import Counter

from datetime import datetime

import string
```

# Reading the Dataset

```python
file_path = "hostel_bois.txt"

with open(file_path, "r", encoding="utf-8") as file:
    lines = file.readlines()

print("="*50)
print("DATASET LOADED")
print("="*50)
print("Total Lines :", len(lines))
```

```python
messages = []

participants = set()

system_messages = 0

media_messages = 0

deleted_messages = 0
```

```python
for line in lines:

    line = line.strip()

    if line == "":
        continue

    # Skip continuation lines (bonus support)
    if " - " not in line:
        continue

    try:

        timestamp, rest = line.split(" - ", 1)

    except ValueError:

        continue

    # System message
    if ": " not in rest:

        system_messages += 1

        continue

    sender, message = rest.split(": ", 1)

    if message == "<Media omitted>":
        media_messages += 1

    if message == "This message was deleted":
        deleted_messages += 1

    participants.add(sender)

    messages.append({

        "timestamp": timestamp,

        "sender": sender,

        "message": message

    })
```

```python
print("="*50)

print("CHAT PARSER SUMMARY")

print("="*50)

print("Messages Parsed :", len(messages))

print("Participants :", len(participants))

print("System Messages :", system_messages)

print("Media Messages :", media_messages)

print("Deleted Messages :", deleted_messages)
```

```python
print(messages[0])
```

```python
print("="*50)

print("PARTICIPANTS")

print("="*50)

for person in sorted(participants):

    print(person)
```

# Group Overview

This section provides an overview of the WhatsApp group by calculating
basic statistics such as total messages, participants, total words,
average words per message, and message contribution by each participant.

```python
message_count = defaultdict(int)
word_count = defaultdict(int)

total_words = 0

for msg in messages:

    sender = msg["sender"]

    text = msg["message"]

    message_count[sender] += 1

    words = text.split()

    word_count[sender] += len(words)

    total_words += len(words)
```

```python
print("=" * 60)
print("GROUP OVERVIEW")
print("=" * 60)

print(f"Total Messages      : {len(messages)}")
print(f"Participants        : {len(participants)}")
print(f"Total Words         : {total_words}")

average_words = total_words / len(messages)

print(f"Average Words/Msg   : {average_words:.2f}")
```

```python
dates = []

for msg in messages:

    timestamp = msg["timestamp"]

    try:

        dt = datetime.strptime(timestamp, "%d/%m/%y, %H:%M")

        dates.append(dt)

    except:

        continue
```

```python
if dates:

    start_date = min(dates)

    end_date = max(dates)

    total_days = (end_date - start_date).days

    print("\nCHAT DURATION")
    print("-" * 40)

    print("Start Date :", start_date.strftime("%d-%m-%Y"))

    print("End Date   :", end_date.strftime("%d-%m-%Y"))

    print("Total Days :", total_days)
```

```python
print("\nMESSAGE CONTRIBUTION")
print("=" * 60)

for user in sorted(message_count):

    count = message_count[user]

    percentage = (count / len(messages)) * 100

    print(f"{user:<20} {count:>5} messages ({percentage:.2f}%)")
```

```python
most_active = max(message_count, key=message_count.get)

least_active = min(message_count, key=message_count.get)

print("\nMOST ACTIVE MEMBER")

print("-" * 40)

print(f"{most_active} ({message_count[most_active]} messages)")

print("\nLEAST ACTIVE MEMBER")

print("-" * 40)

print(f"{least_active} ({message_count[least_active]} messages)")
```

```python
print("\nGROUP OVERVIEW COMPLETED SUCCESSFULLY!")
```

# Time Analysis

This section analyzes when the group is most active by examining
message timestamps. It identifies the busiest hour, busiest day,
and activity distribution across hours and weekdays.

```python
hour_activity = defaultdict(int)
day_activity = defaultdict(int)
weekday_activity = defaultdict(int)

weekday_names = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]
```

```python
for msg in messages:

    try:

        dt = datetime.strptime(msg["timestamp"], "%d/%m/%y, %H:%M")

        hour_activity[dt.hour] += 1

        day_activity[dt.strftime("%d-%m-%Y")] += 1

        weekday_activity[weekday_names[dt.weekday()]] += 1

    except:

        continue
```

```python
print("=" * 60)
print("MESSAGES BY HOUR")
print("=" * 60)

for hour in range(24):

    print(f"{hour:02d}:00 - {hour_activity[hour]} messages")
```

```python
peak_hour = max(hour_activity, key=hour_activity.get)

print("\nMOST ACTIVE HOUR")
print("-" * 40)

print(f"{peak_hour:02d}:00")
print(f"Messages : {hour_activity[peak_hour]}")
```

```python
print("=" * 60)
print("MESSAGES BY WEEKDAY")
print("=" * 60)

for day in weekday_names:

    print(f"{day:<12} : {weekday_activity[day]}")
```

```python
busiest_weekday = max(weekday_activity, key=weekday_activity.get)

print("\nBUSIEST WEEKDAY")
print("-" * 40)

print(busiest_weekday)
print("Messages :", weekday_activity[busiest_weekday])
```

```python
most_active_date = max(day_activity, key=day_activity.get)

print("\nMOST ACTIVE DATE")
print("-" * 40)

print(most_active_date)
print("Messages :", day_activity[most_active_date])
```

```python
print("=" * 60)
print("TOP 10 MOST ACTIVE DAYS")
print("=" * 60)

sorted_days = sorted(
    day_activity.items(),
    key=lambda x: x[1],
    reverse=True
)

for date, count in sorted_days[:10]:

    print(f"{date:<15} {count} messages")
```

# NumPy Activity Heatmap

This section creates a 24-hour activity matrix using NumPy.
Each row represents a participant and each column represents an hour (0–23).

The matrix stores the number of messages sent by each participant during each hour.

```python
users = sorted(list(participants))

user_index = {}

for i, user in enumerate(users):

    user_index[user] = i

print("Participants:")

for i, user in enumerate(users):

    print(i, "-", user)
```

```python
activity_matrix = np.zeros((len(users), 24), dtype=int)

print("Matrix Shape:", activity_matrix.shape)
```

```python
for msg in messages:

    try:

        dt = datetime.strptime(msg["timestamp"], "%d/%m/%y, %H:%M")

        row = user_index[msg["sender"]]

        col = dt.hour

        activity_matrix[row][col] += 1

    except:

        continue
```

```python
print("=" * 90)
print("NUMPY ACTIVITY MATRIX")
print("=" * 90)

print(activity_matrix)
```

```python
print("=" * 90)
print("MESSAGES PER HOUR")
print("=" * 90)

hour_totals = activity_matrix.sum(axis=0)

for hour in range(24):

    print(f"{hour:02d}:00 -> {hour_totals[hour]} messages")
```

```python
print("=" * 90)
print("MESSAGES PER USER")
print("=" * 90)

user_totals = activity_matrix.sum(axis=1)

for i, user in enumerate(users):

    print(f"{user:<20} {user_totals[i]}")
```

```python
overall_peak_hour = np.argmax(hour_totals)

print("\nOVERALL PEAK HOUR")
print("-" * 40)

print(f"{overall_peak_hour:02d}:00")

print("Messages :", hour_totals[overall_peak_hour])
```

```python
print("=" * 90)
print("TEXT HEATMAP")
print("=" * 90)

symbols = [" ", ".", "░", "▒", "▓", "█"]

max_value = activity_matrix.max()

print(f"{'User':<20}", end="")

for h in range(24):

    print(f"{h:02d} ", end="")

print()

for i, user in enumerate(users):

    print(f"{user:<20}", end="")

    for value in activity_matrix[i]:

        if max_value == 0:

            level = 0

        else:

            level = int((value / max_value) * 5)

        print(symbols[level], end="  ")

    print()
```

```python
print("=" * 90)
print("MOST ACTIVE HOUR OF EACH USER")
print("=" * 90)

for i, user in enumerate(users):

    best_hour = np.argmax(activity_matrix[i])

    print(f"{user:<20} {best_hour:02d}:00 ({activity_matrix[i][best_hour]} messages)")
```

```python
print("\nNUMPY HEATMAP COMPLETED SUCCESSFULLY!")
```

# Word Analysis

This section analyzes the textual content of the WhatsApp chat by
counting word frequencies, identifying the most common words,
finding each participant's favorite word, and detecting the
longest and shortest messages.

```python
stop_words = {
    "the","a","an","is","are","to","of","in","on","for","at",
    "and","or","i","you","he","she","it","we","they",
    "me","my","your","our","their","this","that","with",
    "ok","okay","yes","no","ya","bro","da","hi","hello"
}

word_frequency = Counter()

user_word_frequency = defaultdict(Counter)
```

```python
longest_message = ""

shortest_message = None

longest_sender = ""

shortest_sender = ""

translator = str.maketrans("", "", string.punctuation)

for msg in messages:

    sender = msg["sender"]

    text = msg["message"]

    words = text.lower().translate(translator).split()

    filtered_words = []

    for word in words:

        if word not in stop_words and word.strip() != "":

            filtered_words.append(word)

            word_frequency[word] += 1

            user_word_frequency[sender][word] += 1

    if len(text) > len(longest_message):

        longest_message = text

        longest_sender = sender

    if shortest_message is None or len(text) < len(shortest_message):

        shortest_message = text

        shortest_sender = sender
```

```python
print("=" * 70)

print("TOP 20 MOST FREQUENT WORDS")

print("=" * 70)

for word, count in word_frequency.most_common(20):

    print(f"{word:<20} {count}")
```

```python
print("=" * 70)

print("UNIQUE WORDS")

print("=" * 70)

print("Total Unique Words :", len(word_frequency))
```

```python
print("=" * 70)

print("FAVORITE WORD OF EACH PARTICIPANT")

print("=" * 70)

for user in sorted(user_word_frequency):

    if len(user_word_frequency[user]) > 0:

        word, count = user_word_frequency[user].most_common(1)[0]

        print(f"{user:<20} {word:<15} ({count} times)")
```

```python
print("=" * 70)

print("WORD CONTRIBUTION")

print("=" * 70)

for user in sorted(word_count):

    percentage = (word_count[user] / total_words) * 100

    print(f"{user:<20} {word_count[user]:>5} words ({percentage:.2f}%)")
```

```python
print("=" * 70)

print("LONGEST MESSAGE")

print("=" * 70)

print("Sender :", longest_sender)

print("Characters :", len(longest_message))

print()

print(longest_message)
```

```python
print("=" * 70)

print("SHORTEST MESSAGE")

print("=" * 70)

print("Sender :", shortest_sender)

print("Characters :", len(shortest_message))

print()

print(shortest_message)
```

```python
print("=" * 70)

print("WORD ANALYSIS SUMMARY")

print("=" * 70)

print("Unique Words :", len(word_frequency))

print("Most Common Word :", word_frequency.most_common(1)[0][0])

print("Occurrences :", word_frequency.most_common(1)[0][1])

print("\nWORD ANALYSIS COMPLETED SUCCESSFULLY!")
```

# Response Speed & Silent Streaks

This section analyzes how quickly participants respond, identifies
long silent periods in the chat, and assigns personality archetypes
based on chat behavior.

```python
print("="*70)
print("RESPONSE PATTERNS")
print("="*70)

response_times = defaultdict(list)

for i in range(1, len(sorted_messages)):

    previous = sorted_messages[i-1]
    current = sorted_messages[i]

    if previous["sender"] != current["sender"]:

        gap = (current["time"] - previous["time"]).total_seconds()/60

        response_times[current["sender"]].append(gap)

average_response = {}

for user in users:

    if response_times[user]:

        average_response[user] = sum(response_times[user])/len(response_times[user])

fastest = min(average_response, key=average_response.get)

slowest = max(average_response, key=average_response.get)

print(f"Fastest Replier : {fastest} ({average_response[fastest]:.2f} minutes)")
print(f"Slowest Replier : {slowest} ({average_response[slowest]/60:.2f} hours)")
```

```python
for i in range(1, len(sorted_messages)):

    previous = sorted_messages[i - 1]
    current = sorted_messages[i]

    if previous["sender"] != current["sender"]:

        minutes = (current["time"] - previous["time"]).total_seconds() / 60

        if minutes >= 0:

            response_times[current["sender"]].append(minutes)
```

```python
print("=" * 70)
print("AVERAGE RESPONSE TIME")
print("=" * 70)

average_response = {}

for user in sorted(response_times):

    if response_times[user]:

        avg = sum(response_times[user]) / len(response_times[user])

        average_response[user] = avg

        print(f"{user:<20} {avg:.2f} minutes")
```

```python
print("=" * 70)
print("SILENT STREAKS")
print("=" * 70)

largest_gap = 0

gap_start = None
gap_end = None

for i in range(1, len(sorted_messages)):

    diff = (sorted_messages[i]["time"] - sorted_messages[i-1]["time"]).total_seconds()/3600

    if diff > largest_gap:

        largest_gap = diff

        gap_start = sorted_messages[i-1]["time"]

        gap_end = sorted_messages[i]["time"]

print(f"Longest Silent Streak : {largest_gap:.2f} hours")

if gap_start and gap_end:

    print("From :", gap_start.strftime("%d-%m-%Y %H:%M"))

    print("To   :", gap_end.strftime("%d-%m-%Y %H:%M"))
```

```python
print("=" * 70)
print("PERSONALITY ARCHETYPE DETECTION")
print("=" * 70)

archetype_scores = {}

for user in users:
    archetype_scores[user] = {
        "THE SPAMMER": 0,
        "THE GROUP MOM": 0,
        "THE NIGHT OWL": 0,
        "THE STORYTELLER": 0,
        "THE DRAMA QUEEN": 0,
        "THE GHOST": 0,
        "THE COMEDIAN": 0,
        "THE QUESTION MASTER": 0
    }
```

```python
print("=" * 70)
print("PERSONALITY ARCHETYPES")
print("=" * 70)

personality = {}

# ---------------- THE SPAMMER ----------------
bursts = defaultdict(list)

current_sender = None
count = 0

for msg in sorted_messages:

    sender = msg["sender"]

    if sender == current_sender:
        count += 1
    else:
        if current_sender is not None:
            bursts[current_sender].append(count)

        current_sender = sender
        count = 1

if current_sender is not None:
    bursts[current_sender].append(count)

spammer = max(
    users,
    key=lambda u: sum(bursts[u]) / len(bursts[u]) if bursts[u] else 0
)

personality[spammer] = "THE SPAMMER"

# ---------------- THE GROUP MOM ----------------

care_words = [
    "take care",
    "eat",
    "sleep",
    "please",
    "okay",
    "ok",
    "safe",
    "drink water"
]

care_score = defaultdict(int)

for msg in messages:

    text = msg["message"].lower()

    for word in care_words:

        if word in text:

            care_score[msg["sender"]] += 1

remaining = [u for u in users if u not in personality]

group_mom = max(remaining, key=lambda u: care_score[u])

personality[group_mom] = "THE GROUP MOM"

# ---------------- THE NIGHT OWL ----------------

night_score = defaultdict(int)

for msg in messages:

    dt = datetime.strptime(msg["timestamp"], "%d/%m/%y, %H:%M")

    if dt.hour >= 23 or dt.hour <= 4:

        night_score[msg["sender"]] += 1

remaining = [u for u in users if u not in personality]

night_owl = max(remaining, key=lambda u: night_score[u])

personality[night_owl] = "THE NIGHT OWL"

# ---------------- THE STORYTELLER ----------------

avg_words = {}

for user in users:

    texts = [
        m["message"]
        for m in messages
        if m["sender"] == user
    ]

    if texts:

        avg_words[user] = (
            sum(len(t.split()) for t in texts) / len(texts)
        )

    else:

        avg_words[user] = 0

remaining = [u for u in users if u not in personality]

storyteller = max(remaining, key=lambda u: avg_words[u])

personality[storyteller] = "THE STORYTELLER"

# ---------------- THE DRAMA QUEEN ----------------

drama_score = defaultdict(int)

for msg in messages:

    text = msg["message"]

    if text.isupper() and len(text) > 3:

        drama_score[msg["sender"]] += 1

    if text.count("!") >= 2:

        drama_score[msg["sender"]] += 1

remaining = [u for u in users if u not in personality]

drama = max(remaining, key=lambda u: drama_score[u])

personality[drama] = "THE DRAMA QUEEN"

# ---------------- THE GHOST ----------------

remaining = [u for u in users if u not in personality]

ghost = min(remaining, key=lambda u: message_count[u])

personality[ghost] = "THE GHOST"

# ---------------- Remaining Roles ----------------

remaining = [u for u in users if u not in personality]

roles = [
    "THE COMEDIAN",
    "THE QUESTION MASTER"
]

for user, role in zip(remaining, roles):

    personality[user] = role

# ---------------- OUTPUT ----------------

print()

for user in users:

    print(f"{user:<15} : {personality[user]}")
```

```python
print("=" * 70)
print("GROUPDNA - FINAL REPORT")
print("=" * 70)

print(f"Total Participants      : {len(users)}")
print(f"Total Messages          : {len(messages)}")
print(f"System Messages         : {system_messages}")
print(f"Media Messages          : {media_messages}")
print(f"Deleted Messages        : {deleted_messages}")

print("-" * 70)

print(f"Most Active User        : {most_active}")
print(f"Least Active User       : {least_active}")
print(f"Peak Hour               : {peak_hour}:00")
weekday_count = defaultdict(int)

for msg in messages:
    try:
        dt = datetime.strptime(msg["timestamp"], "%d/%m/%y, %H:%M")
        weekday = dt.strftime("%A")
        weekday_count[weekday] += 1
    except Exception:
        pass

busiest_day = max(weekday_count, key=weekday_count.get)

print(f"Busiest Day             : {busiest_day}")

print("-" * 70)

print(f"Fastest Replier         : {fastest}")
print(f"Slowest Replier         : {slowest}")

print("-" * 70)
print()
print("PERSONALITY ARCHETYPES")
print("-" * 70)

for user in users:
    print(f"{user:<18} {personality[user]}")

print("=" * 70)
print("END OF REPORT")
print("=" * 70)
```
